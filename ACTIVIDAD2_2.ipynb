```python
#Importamos librerías requeridas
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Convertir en dataframe
df= pd.read_csv("listings.csv")
df
```

```python
df.info()
df.head()
```

```python
#a) Los registros de 5 “host name”  diferentes
filtro=df["name"].head()
filtro
```

```python
revision1 = df.groupby("name").size().reset_index(name='counts')
revision1
```

```python
#b) Los registros de los host que se hallan unido a Airbnb “despues del año 2020”
filtro1=df[df["host_since"]>"2020-12-31"]
filtro1
```

```python
revision2 = filtro1.groupby("host_since").size().reset_index(name='counts')
revision2
```

```python
#c) Los registros de los host que responden “a más tardar en 1 día”
filtro2=df[~(df["host_response_time"] == "a few days or more")]
filtro2
```

```python
revision3 = filtro2.groupby("host_response_time").size().reset_index(name='counts')
revision3
```

```python
#d) Los registros de las filas “Múltiplos de 200… (400, 600, 800, etc.)”
filtro3 = df.iloc[df.index[df.index % 200 == 0]]
print("Filas múltiplos de 200:")
print(filtro3)
```

```python
revision4 = df.loc[(df.index % 200 == 0) & (df.index != 0)]
print("indices seleccionados:", revision4.index.tolist())
```

```python
#e) Los registros de los tipos de cuarto: “Entire home”  y “Private room”
filtro4=df[(df["beds"] <= 4)]
filtro4
```

```python
revision5=filtro4.groupby("beds").size().reset_index(name='counts')
revision5
```

```python
#f) Los registros que “a lo mucho cuenten con 4 camas” 
filtro5=df[df["beds"] <= 4]
filtro5
```

```python
revision6=filtro5.groupby("beds").size().reset_index(name='counts')
revision6
```

```python
#g) Los registros que “sean superhost”
filtro6 = df[df["host_is_superhost"] == "t"]
filtro6
```

```python
revision7=filtro6.groupby("host_is_superhost").size().reset_index(name='counts')
revision7
```

```python
#h) Los registros de las columnas “impares”
filtro7 = df.iloc[:, 1::2]
filtro7
```

```python
revision8 = filtro7.groupby(filtro7.columns[0]).size().reset_index(name='counts')
revision8
```
